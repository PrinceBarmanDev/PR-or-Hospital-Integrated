```python

import pandas as pd
import numpy as np
import os
import time

INPUT_FILE = r"C:\Users\princ\Dataset Management\openpowerlifting.csv"

OUTPUT_DIR = r"C:\Users\princ\Dataset Management\processed"

os.makedirs(OUTPUT_DIR, exist_ok=True)

SAMPLE_SIZE = 50000
CHUNK_SIZE = 100000

print("PR OR HOSPITAL - DATASET MANAGER")
print("=" * 40)
print("Input:", INPUT_FILE)
print("Output:", OUTPUT_DIR)
print("Target records:", SAMPLE_SIZE)
print("Status: Ready")
```

```python

sample = pd.DataFrame()
rng = np.random.default_rng(42)

total_records = 0
start_time = time.time()

for chunk_number, chunk in enumerate(
    pd.read_csv(
        INPUT_FILE,
        chunksize=CHUNK_SIZE,
        low_memory=False
    ),
    start=1
):
    total_records += len(chunk)

    chunk["_random_key"] = rng.random(len(chunk))

    sample = pd.concat(
        [sample, chunk],
        ignore_index=True
    )

    sample = sample.nsmallest(
        SAMPLE_SIZE,
        "_random_key"
    )

    elapsed = round(time.time() - start_time, 2)

    print(
        f"Chunk {chunk_number} | "
        f"Processed: {total_records:,} | "
        f"Selected: {len(sample):,} | "
        f"Time: {elapsed}s"
    )

sample.drop(columns=["_random_key"], inplace=True)

output_file = os.path.join(
    OUTPUT_DIR,
    "openpowerlifting_50000.csv"
)

sample.to_csv(output_file, index=False)

print("\nSAMPLING COMPLETED!")
print("Total original records:", f"{total_records:,}")
print("Final sampled records:", f"{len(sample):,}")
print("Saved:", output_file)
```

```python

import pandas as pd
import numpy as np
import os

BASE_DIR = r"C:\Users\princ\Dataset Management"

RAW_FILE = os.path.join(
    BASE_DIR, "openpowerlifting.csv"
)

PROCESSED_DIR = os.path.join(BASE_DIR, "processed")
REPORT_DIR = os.path.join(BASE_DIR, "reports")

os.makedirs(PROCESSED_DIR, exist_ok=True)
os.makedirs(REPORT_DIR, exist_ok=True)

print("IPF Dataset Manager initialized!")
```

```python

SAMPLE_FILE = os.path.join(
    PROCESSED_DIR,
    "openpowerlifting_50000.csv"
)

df = pd.read_csv(SAMPLE_FILE)

print("Dataset loaded successfully!")
print("Rows:", len(df))
print("Columns:", len(df.columns))

display(df.head())
```

```python

missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isnull().sum().values,
    "Missing_Percentage": (
        df.isnull().mean().values * 100
    ).round(2)
})

missing_report = missing_report.sort_values(
    by="Missing_Percentage",
    ascending=False
)

display(missing_report)
```

```python

print("Total records:", len(df))

print("Duplicate records:", df.duplicated().sum())

print("Unique athletes (by name):", df["Name"].nunique())

print("Dataset shape:", df.shape)
```

```python
print("df exists:", "df" in globals())
print("sample exists:", "sample" in globals())
print("clean_df exists:", "clean_df" in globals())
```

```python

selected_columns = [
    "Name",
    "Sex",
    "Age",
    "Event",
    "Equipment",
    "BodyweightKg",
    "WeightClassKg",
    "Best3SquatKg",
    "Best3BenchKg",
    "Best3DeadliftKg",
    "TotalKg",
    "Dots",
    "Wilks",
    "Place",
    "Date",
    "Country",
    "Federation",
    "Tested"
]

# Keep only columns that exist in df
available_columns = [
    col for col in selected_columns
    if col in df.columns
]

clean_df = df[available_columns].copy()

print("clean_df successfully created!")
print("Rows:", len(clean_df))
print("Columns:", len(clean_df.columns))

display(clean_df.head())
```

```python
print(clean_df.shape)
print(clean_df.columns.tolist())
```

```python

recovery_features = {
    "SleepDurationHours": "Float64",
    "SleepQuality": "Float64",
    "MuscleSoreness": "Float64",
    "PerceivedStress": "Float64",
    "RestDays": "Int64",
    "TargetLiftType": "string",
    "PlannedAttemptWeightKg": "Float64",
    "RecentSessionRPE": "Float64",
    "DaysSinceLastHeavySession": "Int64",
    "Current1RMKg": "Float64",
    "IntensityPercentage": "Float64",
    "TrainingExperienceYears": "Float64",
    "PainFlag": "boolean",
    "PreviousSuccessfulKg": "Float64",
    "TargetWeightKg": "Float64"
}

for column, dtype in recovery_features.items():
    if column not in clean_df.columns:
        clean_df[column] = pd.Series(
            pd.NA,
            index=clean_df.index,
            dtype=dtype
        )

print("Group H recovery columns added!")
print("Dataset shape:", clean_df.shape)
```
