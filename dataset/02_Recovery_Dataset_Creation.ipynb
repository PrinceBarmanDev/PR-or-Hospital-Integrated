```python

import pandas as pd
import os

# Dataset folder
BASE_DIR = r"C:\Users\princ\Dataset Management"
PROCESSED_DIR = os.path.join(BASE_DIR, "processed")

os.makedirs(PROCESSED_DIR, exist_ok=True)

# Recovery dataset columns
columns = [
    "AthleteID",
    "SessionDate",
    "Age",
    "Sex",
    "BodyweightKg",
    "TargetLiftType",
    "Current1RMKg",
    "PlannedAttemptWeightKg",
    "IntensityPercentage",
    "SleepDurationHours",
    "SleepQuality",
    "MuscleSoreness",
    "PerceivedStress",
    "RecentSessionRPE",
    "RestDays",
    "DaysSinceLastHeavySession",
    "TrainingExperienceYears",
    "PainFlag",
    "PreviousSuccessfulKg",
    "TargetBodyweightKg"
]

# Create an empty dataset
recovery_df = pd.DataFrame(columns=columns)

# Save CSV
RECOVERY_FILE = os.path.join(
    PROCESSED_DIR,
    "recovery_sessions.csv"
)

recovery_df.to_csv(RECOVERY_FILE, index=False)

print("Recovery dataset created successfully!")
print("Number of columns:", len(recovery_df.columns))
print("Saved at:", RECOVERY_FILE)

display(recovery_df.head())
```

```python

# Calculate intensity automatically

recovery_df["IntensityPercentage"] = (
    recovery_df["PlannedAttemptWeightKg"]
    / recovery_df["Current1RMKg"]
) * 100

# Round to 2 decimal places
recovery_df["IntensityPercentage"] = (
    recovery_df["IntensityPercentage"].round(2)
)

display(
    recovery_df[
        [
            "AthleteID",
            "TargetLiftType",
            "Current1RMKg",
            "PlannedAttemptWeightKg",
            "IntensityPercentage"
        ]
    ]
)
```

```python

print("Total recovery sessions:", len(recovery_df))
print("Total features:", len(recovery_df.columns))

print("\nMissing values:")
print(recovery_df.isnull().sum())

print("\nData types:")
print(recovery_df.dtypes)
```

```python

# ==========================================
# STEP: DATA TYPE CONVERSION
# ==========================================

# Numeric columns
float_columns = [
    "Age",
    "BodyweightKg",
    "Current1RMKg",
    "PlannedAttemptWeightKg",
    "IntensityPercentage",
    "SleepDurationHours",
    "TrainingExperienceYears",
    "PreviousSuccessfulKg",
    "TargetBodyweightKg"
]

integer_columns = [
    "SleepQuality",
    "MuscleSoreness",
    "PerceivedStress",
    "RecentSessionRPE",
    "RestDays",
    "DaysSinceLastHeavySession"
]

# Convert numeric columns
for col in float_columns:
    recovery_df[col] = pd.to_numeric(
        recovery_df[col], errors="coerce"
    ).astype("Float64")

for col in integer_columns:
    recovery_df[col] = pd.to_numeric(
        recovery_df[col], errors="coerce"
    ).astype("Int64")

# Convert date
recovery_df["SessionDate"] = pd.to_datetime(
    recovery_df["SessionDate"],
    errors="coerce"
)

# Convert Boolean
recovery_df["PainFlag"] = recovery_df["PainFlag"].astype("boolean")

# Convert text columns
text_columns = [
    "AthleteID",
    "Sex",
    "TargetLiftType"
]

for col in text_columns:
    recovery_df[col] = recovery_df[col].astype("string")

print("Data type conversion completed!")

display(recovery_df.dtypes)
```

```python

# ==========================================
# RECOVERY DATA VALIDATION
# ==========================================

validation_rules = {
    "Age": (14, 100),
    "BodyweightKg": (20, 300),
    "SleepDurationHours": (0, 24),
    "SleepQuality": (1, 10),
    "MuscleSoreness": (1, 10),
    "PerceivedStress": (1, 10),
    "RecentSessionRPE": (1, 10),
    "RestDays": (0, 30),
    "DaysSinceLastHeavySession": (0, 365),
    "Current1RMKg": (1, 1000),
    "PlannedAttemptWeightKg": (1, 1000),
    "TrainingExperienceYears": (0, 80),
    "PreviousSuccessfulKg": (0, 1000),
    "TargetBodyweightKg": (20, 300)
}

for column, (minimum, maximum) in validation_rules.items():

    invalid = (
        recovery_df[column].notna()
        & (
            (recovery_df[column] < minimum)
            | (recovery_df[column] > maximum)
        )
    )

    print(
        f"{column}: {invalid.sum()} invalid records"
    )

print("\nValidation check completed!")
```

```python
recovery_df.to_csv(RECOVERY_FILE, index=False)

print("Recovery dataset saved successfully!")
print("Total sessions:", len(recovery_df))
print("Total features:", len(recovery_df.columns))
print("Saved location:", RECOVERY_FILE)
```
