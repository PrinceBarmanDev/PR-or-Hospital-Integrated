```python

# =====================================================
# PR OR HOSPITAL — RECOVERY DATA COLLECTION
# Notebook 03
# =====================================================

import pandas as pd
import numpy as np
import os
from datetime import datetime

# Project directories
BASE_DIR = r"C:\Users\princ\Dataset Management"

PROCESSED_DIR = os.path.join(
    BASE_DIR,
    "processed"
)

os.makedirs(PROCESSED_DIR, exist_ok=True)

# Recovery dataset path
RECOVERY_FILE = os.path.join(
    PROCESSED_DIR,
    "recovery_sessions.csv"
)

print("PR OR HOSPITAL — RECOVERY DATA COLLECTION")
print("=" * 50)

print("Project folder:", BASE_DIR)
print("Processed folder:", PROCESSED_DIR)
print("Recovery file:", RECOVERY_FILE)

print("\nSetup completed successfully!")
```

```python

# Load existing recovery dataset

if os.path.exists(RECOVERY_FILE):

    recovery_df = pd.read_csv(
        RECOVERY_FILE,
        parse_dates=["SessionDate"]
    )

    print("Existing recovery dataset loaded!")

else:

    recovery_df = pd.DataFrame()

    print("No existing recovery dataset found.")

print("Total sessions:", len(recovery_df))
print("Total columns:", len(recovery_df.columns))

display(recovery_df.head())
```

```python

# =====================================================
# STEP 5 — ATHLETE RECOVERY SESSION INPUT
# =====================================================

def collect_recovery_session():

    print("\n===== ATHLETE RECOVERY SESSION =====\n")

    session = {}

    session["AthleteID"] = input("Athlete ID: ").strip().upper()

    session["SessionDate"] = datetime.now().strftime("%Y-%m-%d")

    session["Age"] = float(input("Age: "))

    session["Sex"] = input("Sex (M/F): ").strip().upper()

    session["BodyweightKg"] = float(
        input("Bodyweight (kg): ")
    )

    session["TargetLiftType"] = input(
        "Target Lift (Squat/Bench/Deadlift): "
    ).strip().title()

    session["Current1RMKg"] = float(
        input("Current 1RM (kg): ")
    )

    session["PlannedAttemptWeightKg"] = float(
        input("Planned Attempt Weight (kg): ")
    )

    session["IntensityPercentage"] = round(
        session["PlannedAttemptWeightKg"]
        / session["Current1RMKg"] * 100,
        2
    )

    session["SleepDurationHours"] = float(
        input("Sleep Duration (hours): ")
    )

    session["SleepQuality"] = int(
        input("Sleep Quality (1-10): ")
    )

    session["MuscleSoreness"] = int(
        input("Muscle Soreness (1-10): ")
    )

    session["PerceivedStress"] = int(
        input("Perceived Stress (1-10): ")
    )

    session["RecentSessionRPE"] = int(
        input("Recent Session RPE (1-10): ")
    )

    session["RestDays"] = int(
        input("Rest Days: ")
    )

    session["DaysSinceLastHeavySession"] = int(
        input("Days Since Last Heavy Session: ")
    )

    session["TrainingExperienceYears"] = float(
        input("Training Experience (years): ")
    )

    pain = input(
        "Joint/Tendon Pain? (yes/no): "
    ).strip().lower()

    session["PainFlag"] = pain in ["yes", "y"]

    session["PreviousSuccessfulKg"] = float(
        input("Previous Successful Weight (kg): ")
    )

    session["TargetBodyweightKg"] = float(
        input("Target Bodyweight (kg): ")
    )

    return session

print("Athlete recovery input function created!")
```

```python

# STEP 6 — RECOVERY INPUT VALIDATION

def validate_recovery_session(session):

    errors = []

    if not session["AthleteID"]:
        errors.append("Athlete ID is required.")

    if not 14 <= session["Age"] <= 100:
        errors.append("Age must be between 14 and 100.")

    if session["Sex"] not in ["M", "F"]:
        errors.append("Sex must be M or F.")

    if not 20 <= session["BodyweightKg"] <= 300:
        errors.append("Invalid bodyweight.")

    if session["TargetLiftType"] not in [
        "Squat", "Bench", "Deadlift"
    ]:
        errors.append("Invalid target lift.")

    if session["Current1RMKg"] <= 0:
        errors.append("Current 1RM must be positive.")

    if session["PlannedAttemptWeightKg"] <= 0:
        errors.append("Planned weight must be positive.")

    if not 0 <= session["SleepDurationHours"] <= 24:
        errors.append("Invalid sleep duration.")

    for field in [
        "SleepQuality",
        "MuscleSoreness",
        "PerceivedStress",
        "RecentSessionRPE"
    ]:
        if not 1 <= session[field] <= 10:
            errors.append(f"{field} must be between 1 and 10.")

    if session["RestDays"] < 0:
        errors.append("Rest days cannot be negative.")

    if session["DaysSinceLastHeavySession"] < 0:
        errors.append("Days since last heavy session cannot be negative.")

    if session["TrainingExperienceYears"] < 0:
        errors.append("Training experience cannot be negative.")

    if session["PreviousSuccessfulKg"] < 0:
        errors.append("Previous successful weight cannot be negative.")

    if not 20 <= session["TargetBodyweightKg"] <= 300:
        errors.append("Invalid target bodyweight.")

    if errors:
        return False, errors

    return True, []

print("Recovery validation function created!")
```

```python

# STEP 7 — TEST RECOVERY VALIDATION

test_session = {
    "AthleteID": "ATH003",
    "SessionDate": datetime.now().strftime("%Y-%m-%d"),
    "Age": 25.0,
    "Sex": "M",
    "BodyweightKg": 82.0,
    "TargetLiftType": "Squat",
    "Current1RMKg": 190.0,
    "PlannedAttemptWeightKg": 155.0,
    "IntensityPercentage": 81.58,
    "SleepDurationHours": 8.0,
    "SleepQuality": 8,
    "MuscleSoreness": 3,
    "PerceivedStress": 4,
    "RecentSessionRPE": 7,
    "RestDays": 2,
    "DaysSinceLastHeavySession": 4,
    "TrainingExperienceYears": 4.0,
    "PainFlag": False,
    "PreviousSuccessfulKg": 150.0,
    "TargetBodyweightKg": 82.0
}

valid, errors = validate_recovery_session(test_session)

print("Valid:", valid)
print("Errors:", errors)
```

```python

# STEP 8 — SAVE RECOVERY SESSION

def save_recovery_session(session):

    global recovery_df

    # Validate session
    valid, errors = validate_recovery_session(session)

    if not valid:
        print("❌ Session validation failed:")
        for error in errors:
            print("-", error)
        return

    # Convert session into a DataFrame
    new_session_df = pd.DataFrame([session])

    # Match existing column order
    new_session_df = new_session_df[recovery_df.columns]

    # Add new session to existing records
    updated_df = pd.concat(
        [recovery_df, new_session_df],
        ignore_index=True
    )

    # Save safely using a temporary file
    TEMP_FILE = RECOVERY_FILE + ".tmp"

    updated_df.to_csv(TEMP_FILE, index=False)
    os.replace(TEMP_FILE, RECOVERY_FILE)

    # Update the in-memory DataFrame
    recovery_df = updated_df

    print("✅ Recovery session saved successfully!")
    print("Total sessions:", len(recovery_df))
```
